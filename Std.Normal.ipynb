```python
import numpy as np
from scipy.stats import norm   # norm with default loc=0, scale=1 IS the standard normal

# CDF and its inverse
print(norm.cdf(1.96))     # 0.9750
print(norm.ppf(0.975))    # 1.96   (inverse: percentile -> z)
print(norm.ppf(0.5))      # 0.0

# Symmetry check: Phi(-z) == 1 - Phi(z)
z = 1.3
print(norm.cdf(-z), 1 - norm.cdf(z))   # equal

# Standardizing real data (the z-score transform)
x = np.array([1050, 1200, 1350, 1650])      # coffee-shop revenue samples
mu, sigma = 1200, 150
z = (x - mu) / sigma
print(z)                  # [-1.  0.  1.  3.]  -> SDs from the mean

# ...which is what sklearn.preprocessing.StandardScaler does column-wise
```

```python
import matplotlib.pyplot as plt

# Standard Normal Distribution: mean=0, std=1
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x)

fig, ax = plt.subplots(figsize=(8, 5))

BLUE = "#2a78d6"

# Shade ±1, ±2, ±3 SD bands (68-95-99.7 rule), lightest first
for k, alpha in zip([3, 2, 1], [0.12, 0.20, 0.30]):
    band = (x >= -k) & (x <= k)
    ax.fill_between(x[band], y[band], color=BLUE, alpha=alpha, linewidth=0)

ax.plot(x, y, color=BLUE, linewidth=2)
ax.axvline(0, color="#52514e", linewidth=1, linestyle="--", alpha=0.6)

ax.set_title("Standard Normal Distribution  (μ=0, σ=1)")
ax.set_xlabel("z")
ax.set_ylabel("Probability Density")
ax.set_xticks(range(-4, 5))
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```
