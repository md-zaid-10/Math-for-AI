```python
import numpy as np
import matplotlib.pyplot as plt

# Power Law Distribution: p(x) = C * x^(-alpha)
alpha = 2.5   # power law exponent

x = np.linspace(1, 50, 1000)
y = x ** (-alpha)

plt.plot(x, y, color="#d62a2a", linewidth=2)
plt.title("Power Law Distribution (alpha=2.5)")
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.show()
```
