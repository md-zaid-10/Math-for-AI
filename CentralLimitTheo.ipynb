## Central Limit Theorem

No matter the shape of the original population, the distribution of sample means approaches a normal distribution as the sample size grows.

```python
import numpy as np
import matplotlib.pyplot as plt

population = np.random.exponential(scale=2.0, size=100000)  # skewed population
sample_size = 30
num_samples = 5000

sample_means = [np.mean(np.random.choice(population, sample_size)) for _ in range(num_samples)]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(population, bins=50, color='orange', alpha=0.7)
axes[0].set_title("Original Population (Exponential)")

axes[1].hist(sample_means, bins=50, color='green', alpha=0.7)
axes[1].set_title(f"Sample Means (n={sample_size})")

plt.tight_layout()
plt.show()
```
