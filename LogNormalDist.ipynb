```python
import numpy as np
from scipy.stats import lognorm
import matplotlib.pyplot as plt

# Log-Normal Distribution: log(X) ~ Normal(mean, sigma)
sigma = 0.5   # shape parameter (std dev of log(X))

x = np.linspace(0.01, 5, 1000)
y = lognorm.pdf(x, sigma)

plt.plot(x, y, color="#2a78d6", linewidth=2)
plt.title("Log-Normal Distribution (sigma=0.5)")
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.show()
```
